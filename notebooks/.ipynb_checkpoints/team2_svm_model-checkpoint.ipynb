```python
# =============================================================================
# TEAM 2: SVM Model Development & Evaluation
# Currency Exchange Movement Prediction
# =============================================================================
# 
# Project: USD Exchange Rate vs Top Currencies
# Task: Predict if EUR/USD will go UP (1) or DOWN/EQUAL (0) tomorrow
# 
# Author: Team 2
# Date: 2026-09-08
# =============================================================================

import os
import sys
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    confusion_matrix,
    classification_report
)
from sklearn.inspection import permutation_importance

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# Set visualization style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

# =============================================================================
# CONFIGURATION
# =============================================================================

CONFIG = {
    "data_path": "../data/processed/processed_data.csv",
    "results_dir": "../results",
    "plots_dir": "../results/plots",
    "metrics_dir": "../results/metrics",
    "model_dir": "../models",
    "test_size": 0.2,
    "random_state": 42,
    "target_column": "target",
    "feature_columns": ["lag_1", "lag_2", "lag_3", "return_1d", "ma_5", "ma_10"],
    "param_grid": {
        "C": [0.1, 1, 10, 100],
        "gamma": ["scale", "auto", 0.01, 0.1],
        "kernel": ["rbf", "linear"],
        "class_weight": ["balanced", None]
    }
}

# =============================================================================
# 1. SETUP DIRECTORIES
# =============================================================================

def setup_directories():
    """Create necessary directories for outputs."""
    dirs = [
        CONFIG["results_dir"],
        CONFIG["plots_dir"],
        CONFIG["metrics_dir"],
        CONFIG["model_dir"]
    ]
    for d in dirs:
        os.makedirs(d, exist_ok=True)
    print("✓ Directories created successfully")

setup_directories()

# =============================================================================
# 2. DATA LOADING
# =============================================================================

class DataLoader:
    """
    Handles loading and initial exploration of the processed dataset.
    
    Team 2 expects the exact column structure agreed upon:
    - Features: lag_1, lag_2, lag_3, return_1d, ma_5, ma_10
    - Target: target (binary: 0 = down/equal, 1 = up)
    """
    
    def __init__(self, data_path):
        self.data_path = data_path
        self.data = None
        self.X = None
        self.y = None
        
    def load_data(self):
        """Load the processed CSV file."""
        try:
            self.data = pd.read_csv(self.data_path)
            print(f"✓ Data loaded successfully: {self.data.shape[0]} rows, {self.data.shape[1]} columns")
            return self.data
        except FileNotFoundError:
            print(f"❌ Error: File not found at {self.data_path}")
            raise
    
    def explore(self):
        """Perform initial data exploration."""
        print("\n" + "="*70)
        print("DATA EXPLORATION")
        print("="*70)
        
        print("\n📊 Dataset Overview:")
        print(f"   - Shape: {self.data.shape}")
        print(f"   - Columns: {list(self.data.columns)}")
        
        print("\n📈 Feature Statistics:")
        print(self.data.describe())
        
        print("\n🎯 Target Distribution:")
        target_counts = self.data['target'].value_counts()
        print(target_counts)
        print(f"   - Class 0 (Down/Equal): {target_counts[0]} ({target_counts[0]/len(self.data)*100:.1f}%)")
        print(f"   - Class 1 (Up): {target_counts[1]} ({target_counts[1]/len(self.data)*100:.1f}%)")
        
        print("\n🔍 Missing Values:")
        print(self.data.isnull().sum())
        
        return self.data
    
    def split_features_target(self):
        """Split data into features (X) and target (y)."""
        self.X = self.data[CONFIG["feature_columns"]].copy()
        self.y = self.data[CONFIG["target_column"]].copy()
        print(f"\n✓ Features shape: {self.X.shape}")
        print(f"✓ Target shape: {self.y.shape}")
        return self.X, self.y

# =============================================================================
# 3. DATA PREPARATION (Train/Test Split & Scaling)
# =============================================================================

class DataPreparer:
    """
    Handles train/test split (time-series aware) and feature scaling.
    
    IMPORTANT: No random shuffle - we respect time order since this is
    time-series data.
    """
    
    def __init__(self, X, y, test_size=0.2):
        self.X = X
        self.y = y
        self.test_size = test_size
        self.scaler = StandardScaler()
        self.X_train = None
        self.X_test = None
        self.y_train = None
        self.y_test = None
        self.X_train_scaled = None
        self.X_test_scaled = None
        
    def split_data(self):
        """Split data chronologically (no shuffle)."""
        split_idx = int(len(self.X) * (1 - self.test_size))
        
        self.X_train = self.X.iloc[:split_idx]
        self.X_test = self.X.iloc[split_idx:]
        self.y_train = self.y.iloc[:split_idx]
        self.y_test = self.y.iloc[split_idx:]
        
        print("\n" + "="*70)
        print("TRAIN/TEST SPLIT (Time-Series Aware)")
        print("="*70)
        print(f"   - Training samples: {len(self.X_train)} ({100*(1-self.test_size):.0f}%)")
        print(f"   - Testing samples: {len(self.X_test)} ({100*self.test_size:.0f}%)")
        print(f"   - Train period: {self.X_train.index[0]} to {self.X_train.index[-1]}")
        print(f"   - Test period: {self.X_test.index[0]} to {self.X_test.index[-1]}")
        
        return self.X_train, self.X_test, self.y_train, self.y_test
    
    def scale_features(self):
        """
        Scale features using StandardScaler.
        Fit on training data only, transform both train and test.
        """
        self.X_train_scaled = self.scaler.fit_transform(self.X_train)
        self.X_test_scaled = self.scaler.transform(self.X_test)
        
        print("\n✓ Feature scaling completed")
        print(f"   - Scaler: StandardScaler")
        print(f"   - Training mean: {self.scaler.mean_.round(3)}")
        print(f"   - Training std: {self.scaler.scale_.round(3)}")
        
        return self.X_train_scaled, self.X_test_scaled
    
    def get_data(self):
        """Return all prepared data."""
        return {
            'X_train': self.X_train_scaled,
            'X_test': self.X_test_scaled,
            'y_train': self.y_train,
            'y_test': self.y_test,
            'scaler': self.scaler,
            'feature_names': list(self.X.columns)
        }

# =============================================================================
# 4. SVM MODEL TRAINING
# =============================================================================

class SVMModel:
    """
    SVM model with hyperparameter optimization using GridSearchCV.
    
    Uses TimeSeriesSplit for cross-validation to respect temporal order.
    """
    
    def __init__(self, param_grid=None):
        self.param_grid = param_grid or CONFIG["param_grid"]
        self.model = None
        self.best_params = None
        self.best_score = None
        self.cv_results = None
        
    def train(self, X_train, y_train):
        """
        Train SVM with hyperparameter tuning.
        
        Args:
            X_train: Training features (scaled)
            y_train: Training target
        """
        print("\n" + "="*70)
        print("SVM MODEL TRAINING")
        print("="*70)
        
        # Use TimeSeriesSplit for time-series cross-validation
        tscv = TimeSeriesSplit(n_splits=5)
        
        # Initialize GridSearchCV
        grid_search = GridSearchCV(
            estimator=SVC(random_state=CONFIG["random_state"]),
            param_grid=self.param_grid,
            cv=tscv,
            scoring='f1',
            n_jobs=-1,
            verbose=1
        )
        
        print(f"\n🔍 Hyperparameter Search:")
        print(f"   - Parameter grid: {self.param_grid}")
        print(f"   - CV folds: {tscv.n_splits}")
        print(f"   - Scoring metric: f1")
        print(f"   - Total fits: {grid_search.n_splits_} CV × {sum(len(v) for v in self.param_grid.values())} params")
        
        # Train
        grid_search.fit(X_train, y_train)
        
        # Store results
        self.model = grid_search.best_estimator_
        self.best_params = grid_search.best_params_
        self.best_score = grid_search.best_score_
        self.cv_results = grid_search.cv_results_
        
        print(f"\n✅ Best Parameters: {self.best_params}")
        print(f"✅ Best CV F1 Score: {self.best_score:.4f}")
        
        return self.model
    
    def predict(self, X_test):
        """Make predictions on test data."""
        return self.model.predict(X_test)
    
    def predict_proba(self, X_test):
        """Get probability predictions."""
        return self.model.predict_proba(X_test)

# =============================================================================
# 5. MODEL EVALUATION
# =============================================================================

class ModelEvaluator:
    """
    Comprehensive model evaluation with metrics and visualizations.
    """
    
    def __init__(self, model, X_test, y_test, feature_names):
        self.model = model
        self.X_test = X_test
        self.y_test = y_test
        self.feature_names = feature_names
        self.y_pred = None
        self.y_proba = None
        self.metrics = {}
        
    def evaluate(self):
        """Run full evaluation pipeline."""
        print("\n" + "="*70)
        print("MODEL EVALUATION")
        print("="*70)
        
        # Get predictions
        self.y_pred = self.model.predict(self.X_test)
        self.y_proba = self.model.predict_proba(self.X_test)[:, 1]
        
        # Calculate metrics
        self.metrics = {
            'accuracy': accuracy_score(self.y_test, self.y_pred),
            'precision': precision_score(self.y_test, self.y_pred),
            'recall': recall_score(self.y_test, self.y_pred),
            'f1': f1_score(self.y_test, self.y_pred),
            'roc_auc': roc_auc_score(self.y_test, self.y_proba)
        }
        
        # Print metrics
        print("\n📊 Performance Metrics:")
        for metric, value in self.metrics.items():
            print(f"   - {metric.upper()}: {value:.4f}")
        
        # Classification report
        print("\n📋 Classification Report:")
        print(classification_report(self.y_test, self.y_pred, 
                                     target_names=['Down/Equal (0)', 'Up (1)']))
        
        return self.metrics
    
    def plot_confusion_matrix(self, save_path=None):
        """Plot and save confusion matrix."""
        cm = confusion_matrix(self.y_test, self.y_pred)
        
        fig, ax = plt.subplots(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                    xticklabels=['Down/Equal (0)', 'Up (1)'],
                    yticklabels=['Down/Equal (0)', 'Up (1)'],
                    ax=ax)
        ax.set_title('Confusion Matrix', fontsize=14, fontweight='bold')
        ax.set_xlabel('Predicted', fontsize=12)
        ax.set_ylabel('Actual', fontsize=12)
        
        plt.tight_layout()
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"✓ Confusion matrix saved to: {save_path}")
        
        plt.show()
        return fig
    
    def plot_roc_curve(self, save_path=None):
        """Plot and save ROC curve."""
        fpr, tpr, thresholds = roc_curve(self.y_test, self.y_proba)
        
        fig, ax = plt.subplots(figsize=(8, 6))
        
        # Plot ROC curve
        ax.plot(fpr, tpr, linewidth=2, label=f'ROC (AUC = {self.metrics["roc_auc"]:.4f})')
        ax.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random Classifier')
        
        ax.set_xlim([0.0, 1.0])
        ax.set_ylim([0.0, 1.05])
        ax.set_xlabel('False Positive Rate', fontsize=12)
        ax.set_ylabel('True Positive Rate', fontsize=12)
        ax.set_title('ROC Curve', fontsize=14, fontweight='bold')
        ax.legend(loc='lower right', fontsize=11)
        ax.grid(True, alpha=0.3)
        
        plt.tight_layout()
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"✓ ROC curve saved to: {save_path}")
        
        plt.show()
        return fig
    
    def plot_feature_importance(self, save_path=None):
        """
        Calculate and plot permutation feature importance.
        """
        print("\n" + "="*70)
        print("FEATURE IMPORTANCE (Permutation Importance)")
        print("="*70)
        
        # Calculate permutation importance
        result = permutation_importance(
            self.model, self.X_test, self.y_test,
            n_repeats=10, random_state=CONFIG["random_state"],
            scoring='f1'
        )
        
        # Create DataFrame
        importance_df = pd.DataFrame({
            'feature': self.feature_names,
            'importance': result.importances_mean,
            'std': result.importances_std
        }).sort_values('importance', ascending=True)
        
        # Print
        print("\n📊 Feature Importance:")
        for _, row in importance_df.sort_values('importance', ascending=False).iterrows():
            print(f"   - {row['feature']}: {row['importance']:.4f} (±{row['std']:.4f})")
        
        # Plot
        fig, ax = plt.subplots(figsize=(10, 6))
        bars = ax.barh(importance_df['feature'], importance_df['importance'],
                       xerr=importance_df['std'], capsize=4,
                       color=sns.color_palette('husl', len(importance_df)))
        ax.set_xlabel('Permutation Importance (F1 drop)', fontsize=12)
        ax.set_title('Feature Importance', fontsize=14, fontweight='bold')
        ax.grid(True, alpha=0.3, axis='x')
        
        # Add value labels
        for i, (v, std) in enumerate(zip(importance_df['importance'], importance_df['std'])):
            ax.text(v + std + 0.001, i, f'{v:.4f}', va='center', fontsize=10)
        
        plt.tight_layout()
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"✓ Feature importance plot saved to: {save_path}")
        
        plt.show()
        return fig, importance_df
    
    def plot_predictions(self, save_path=None):
        """Plot actual vs predicted over time."""
        # Convert to numpy if needed
        y_test_np = self.y_test.values if hasattr(self.y_test, 'values') else np.array(self.y_test)
        
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10))
        
        # Plot 1: Actual vs Predicted
        ax1.plot(range(len(y_test_np)), y_test_np, 'b-', alpha=0.7, label='Actual', linewidth=1)
        ax1.plot(range(len(y_test_np)), self.y_pred, 'r-', alpha=0.5, label='Predicted', linewidth=0.8)
        ax1.set_xlabel('Test Sample Index', fontsize=12)
        ax1.set_ylabel('Movement Direction', fontsize=12)
        ax1.set_title('Actual vs Predicted (Test Set)', fontsize=14, fontweight='bold')
        ax1.legend(loc='upper right')
        ax1.grid(True, alpha=0.3)
        
        # Plot 2: Probability distribution
        ax2.hist(self.y_proba[self.y_test == 0], bins=20, alpha=0.6, label='Actual Down/Equal (0)', color='blue')
        ax2.hist(self.y_proba[self.y_test == 1], bins=20, alpha=0.6, label='Actual Up (1)', color='red')
        ax2.set_xlabel('Predicted Probability (Class 1)', fontsize=12)
        ax2.set_ylabel('Frequency', fontsize=12)
        ax2.set_title('Predicted Probability Distribution', fontsize=14, fontweight='bold')
        ax2.legend(loc='upper center')
        ax2.grid(True, alpha=0.3)
        
        plt.tight_layout()
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"✓ Predictions plot saved to: {save_path}")
        
        plt.show()
        return fig
    
    def save_metrics(self, save_path=None):
        """Save metrics to JSON file."""
        if save_path is None:
            save_path = os.path.join(CONFIG["metrics_dir"], "metrics.json")
        
        # Add timestamp and model info
        full_metrics = {
            'timestamp': datetime.now().isoformat(),
            'model': 'SVM',
            'best_params': self.model.get_params(),
            'metrics': self.metrics,
            'test_samples': len(self.y_test)
        }
        
        with open(save_path, 'w') as f:
            json.dump(full_metrics, f, indent=2)
        
        print(f"\n✓ Metrics saved to: {save_path}")
        return full_metrics

# =============================================================================
# 6. MAIN PIPELINE
# =============================================================================

def run_pipeline():
    """
    Execute the complete Team 2 pipeline.
    """
    print("\n" + "="*70)
    print(" " * 20 + "TEAM 2: SVM PIPELINE")
    print("="*70)
    print(f"Started: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    
    # Step 1: Load data
    loader = DataLoader(CONFIG["data_path"])
    data = loader.load_data()
    loader.explore()
    X, y = loader.split_features_target()
    
    # Step 2: Prepare data
    preparer = DataPreparer(X, y, test_size=CONFIG["test_size"])
    preparer.split_data()
    preparer.scale_features()
    prepared_data = preparer.get_data()
    
    # Step 3: Train model
    model_trainer = SVMModel()
    model = model_trainer.train(
        prepared_data['X_train'],
        prepared_data['y_train']
    )
    
    # Step 4: Evaluate model
    evaluator = ModelEvaluator(
        model=model,
        X_test=prepared_data['X_test'],
        y_test=prepared_data['y_test'],
        feature_names=prepared_data['feature_names']
    )
    
    metrics = evaluator.evaluate()
    
    # Step 5: Generate visualizations
    print("\n" + "="*70)
    print("VISUALIZATIONS")
    print("="*70)
    
    # Confusion Matrix
    cm_path = os.path.join(CONFIG["plots_dir"], "confusion_matrix.png")
    evaluator.plot_confusion_matrix(save_path=cm_path)
    
    # ROC Curve
    roc_path = os.path.join(CONFIG["plots_dir"], "roc_curve.png")
    evaluator.plot_roc_curve(save_path=roc_path)
    
    # Feature Importance
    fi_path = os.path.join(CONFIG["plots_dir"], "feature_importance.png")
    _, importance_df = evaluator.plot_feature_importance(save_path=fi_path)
    
    # Predictions Plot
    pred_path = os.path.join(CONFIG["plots_dir"], "predictions.png")
    evaluator.plot_predictions(save_path=pred_path)
    
    # Step 6: Save metrics
    metrics_path = os.path.join(CONFIG["metrics_dir"], "metrics.json")
    evaluator.save_metrics(save_path=metrics_path)
    
    # Step 7: Summary
    print("\n" + "="*70)
    print(" " * 25 + "SUMMARY")
    print("="*70)
    print(f"\n✅ Pipeline completed successfully!")
    print(f"   - Best F1 Score: {model_trainer.best_score:.4f}")
    print(f"   - Test Accuracy: {metrics['accuracy']:.4f}")
    print(f"   - Test F1 Score: {metrics['f1']:.4f}")
    print(f"   - Test AUC: {metrics['roc_auc']:.4f}")
    
    print("\n📁 Outputs:")
    print(f"   - Plots: {CONFIG['plots_dir']}")
    print(f"   - Metrics: {CONFIG['metrics_dir']}")
    print(f"   - Model: {CONFIG['model_dir']}")
    
    print(f"\nCompleted: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("\n" + "="*70)
    
    return {
        'model': model,
        'scaler': prepared_data['scaler'],
        'metrics': metrics,
        'feature_importance': importance_df,
        'best_params': model_trainer.best_params
    }

# =============================================================================
# 7. EXECUTION
# =============================================================================

if __name__ == "__main__":
    results = run_pipeline()
```
